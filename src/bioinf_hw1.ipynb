## Контиги

Сбор контигов

```python
contig_out_dir = '/content/contig_res/'

!mkdir -p {contig_out_dir}

!./platanus assemble -f {data_dir}*.trimmed -o {contig_out_dir} -t 4 -m 8
```

Анализ контигов

```python
def get_N50(lens):
  sum_len = sum(lens)
  lens.sort(reverse=True)

  cur_sum = 0

  for len in lens:
    cur_sum += len
    if cur_sum * 2 >= sum_len:
      return len


def analise_contigs(contig_file):
  lens = []
  cur_len = 0

  with open(contig_file, 'r') as f:
    for line in f:
      clean_line = line.strip()
      if clean_line[0] == '>':
        if cur_len > 0:
          lens.append(cur_len)
        cur_len = 0
      else:
        cur_len += len(clean_line)

    if cur_len > 0:
        lens.append(cur_len)


  print(f'Число контигов/скаффолдов = {len(lens)}')
  print(f'Суммарная длина = {sum(lens)}')
  print(f'Максимальная длина контига = {max(lens)}')
  print(f'N50 = {get_N50(lens)}')


contig_main = f'{contig_out_dir}/_contig.fa'

analise_contigs(contig_main)
```

## Скаффолды

Выделим скаффолды

```python
scaffold_out_dir = '/content/scaffold_res/'

!mkdir -p {scaffold_out_dir}

contig_bubble = f'{contig_out_dir}/_contigBubble.fa'

!./platanus scaffold -c {contig_main} -b {contig_bubble} -IP1 {PE_1} {PE_2} -OP {MP_1} {MP_2} -o {scaffold_out_dir} -t 4
```

Проанализируем

```python
def write_longest_scaffold(id, scaffold, out_file):
  with open(out_file, 'w') as f:
    f.write(f'{id}\n')
    f.write(scaffold)

def analise_longest_scaffold(scaffold_file, out_file):
  longest_scaffold = ''
  cur_scaffold = ''

  best_id = ''
  cur_id = ''

  best_len = 0
  cur_len = 0

  with open(scaffold_file, 'r') as f:
    for line in f:
      clean_line = line.strip()
      if clean_line[0] == '>':
        if cur_len > best_len:
          longest_scaffold = cur_scaffold
          best_id = cur_id
          best_len = cur_len
        cur_len = 0
        cur_id = clean_line
        cur_scaffold = ''
      else:
        cur_scaffold += clean_line
        cur_len += len(clean_line)

    if cur_len > best_len:
        longest_scaffold = cur_scaffold
        best_id = cur_id
        best_len = cur_len

  cnt_gaps = 0
  len_gaps = 0
  mark = 0

  for l in longest_scaffold:
    if l == 'N':
      len_gaps += 1
      if mark == 0:
        mark = 1
        cnt_gaps += 1
    else:
      mark = 0

  print(f'Число гэпов в самом длинном скаффолде = {cnt_gaps}')
  print(f'Суммарная длина гэпов в самом длинном скаффолде = {len_gaps}')

  write_longest_scaffold(best_id, longest_scaffold, out_file)

scaffold_main = f'{scaffold_out_dir}/_scaffold.fa'
out_file = f'{scaffold_out_dir}/longest_scaffold.fa'

analise_contigs(scaffold_main)
analise_longest_scaffold(scaffold_main, out_file)
```

Уменьшение числа гэпов

```python
gap_close_main = f'{gap_close_out_dir}/_gapClosed.fa'
out_file = f'{gap_close_out_dir}/longest_scaffold.fa'

analise_longest_scaffold(gap_close_main, out_file)
```

Как видим, гэпов действительно стало меньше - их число уменьшилось в 6 раз, а суммарная длина в 3, результат получился таким, каким мы хотели
